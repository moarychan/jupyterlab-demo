```python
import numpy as np

A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

print("A + B =\n", A + B)
print("A x B =\n", A @ B)
print("det(A) =", np.linalg.det(A))
```

```python
import matplotlib.pyplot as plt

t = np.linspace(0, 24 * np.pi, 12000)
r = np.sin(7 * t) + 0.35 * np.sin(17 * t)
x = r * np.cos(t)
y = r * np.sin(t)

fig, ax = plt.subplots(figsize=(8, 8), facecolor="#050816")
colors = np.linspace(0, 1, len(t) - 1)
for i in range(len(t) - 1):
    ax.plot(x[i:i + 2], y[i:i + 2], color=plt.cm.turbo(colors[i]), linewidth=0.6)

ax.set_aspect("equal")
ax.axis("off")
ax.set_title("Mathematical Flower", color="white", fontsize=18, pad=20)
plt.show()
```
